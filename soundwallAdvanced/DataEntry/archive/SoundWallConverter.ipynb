```python
import json
import pandas as pd

# Load the JSON file with a specific encoding
with open('../data/soundwallBack.json', 'r', encoding='utf-8-sig') as file:
    data = json.load(file)

# Initialize a list to store the extracted data
extracted_words = []

# Define a function to extract "words" data from the given section
def extract_words_data(section):
    for item in data.get(section, []):
        # Ensure 'content' key exists and has at least one item
        if 'content' in item and len(item['content']) > 0:
            for child in item['content'][0].get('children', []):
                if child.get('type') == 'icn':
                    # Find the grp objects within the icn
                    for subchild in child.get('children', []):
                        if subchild.get('type') == 'grp':
                            parent_id = subchild.get('id', 'N/A')  # Default to 'N/A' if 'id' is missing
                            # Extract words from the "txt" objects within the grp
                            for grp_child in subchild.get('children', []):
                                if grp_child.get('type') == 'txt':
                                    word = grp_child.get('content', 'N/A')  # Extract the word content
                                    extracted_words.append([parent_id, word])

# Extract words data from VV2 and CC2 sections
extract_words_data('vowelValley2')
extract_words_data('consonantCanyon2')

# Convert the extracted data into a DataFrame
df_words = pd.DataFrame(extracted_words, columns=['Parent Div ID', 'Word'])

# Save the DataFrame to an Excel file
output_file_words = 'vv2_cc2_words_ladder.xlsx'
df_words.to_excel(output_file_words, index=False)

output_file_words
```
